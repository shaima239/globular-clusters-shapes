# Histograms for All Clusters
After processing all 126 clusters, read the results from a text file and plot histograms of ellipticity and position angle.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import colors
from matplotlib.ticker import FormatStrFormatter
from matplotlib.ticker import NullFormatter

fig, ax =  plt.subplots(figsize=(6,6))

f = open('GCs.txt', "r")
lines = f.readlines()
Ellipticity = []
for x in lines:
    Ellipticity.append(float(x.split(',')[1]))
f.close()   


n, bins, patches = plt.hist(Ellipticity, bins=20, range=(0, 0.4), histtype='step', edgecolor='black', facecolor='darkseagreen',
               alpha=1)
ax.set_xlabel(" Ellipticity ", fontsize=16)
ax.set_ylabel(" Number of Clusters ", fontsize=12)

print(np.mean(Ellipticity))
print(np.std(Ellipticity))
g = (np.std(Ellipticity))/np.sqrt(126)
print(g)
print()
plt.show()


```
